```python
from bs4 import BeautifulSoup
import requests
import time
import datetime
import smtplib
```

```python
## Connecting to the website 

##url of the item on amazon store
URL = "https://www.amazon.co.za/Samsung-Odyssey-Gaming-Monitor-34-Inch/dp/B0BRL8B5L5/ref=sr_1_3?sr=8-3"

headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/147.0.0.0 Safari/537.36"}

page = requests.get(URL, headers = headers)

soup1 = BeautifulSoup(page.content, 'html.parser')

soup2 = BeautifulSoup(soup1.prettify(), 'html.parser')

title =soup2.find(id = 'productTitle').get_text()

today = datetime.date.today()


price = soup2.find(class_ ='a-price-whole').get_text()
price =price.strip()[:5]
title=title.strip()[:60]

print(len(title))
```

```python
import csv

header = ['Product Name', 'Price','Date']
data = [title,price,today]



with open('AmazonWebScrapedDataset.csv','w',newline='', encoding='UTF8') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerow(data)

print(data)
```
